```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', 100)
```

## Load

```python
DATA_SOURCE = 'data/wb_panel_dated.csv'


# The strings read as missing, per column, exactly as the analysis read them.
MISSING_STRINGS = {
    **dict.fromkeys(['country_code', 'country'], ['']),
    **dict.fromkeys(['year', 'life_expectancy', 'health_spend_ppp', 'health_spend_usd', 'gdp_per_capita_ppp', 'population', 'date'], ['', '#N/A', '#N/A N/A', '#NA', '-1.#IND', '-1.#QNAN', '-NaN', '-nan', '1.#IND', '1.#QNAN', '<NA>', 'N/A', 'NA', 'NULL', 'NaN', 'None', 'n/a', 'nan', 'null']),
}


def load() -> pd.DataFrame:
    df = pd.read_csv(
        DATA_SOURCE, sep=',', encoding='utf-8', keep_default_na=False, na_values=MISSING_STRINGS)
    return df.dropna(axis=1, how="all")  # all-empty columns, as the analysis dropped them


df = load()
print(df.shape)
df.head()
```

## Transform (Avaloka-selected)

```python
def transform(df):
    for _c in ['country_code', 'country', 'date']:
        df[_c] = df[_c].astype('string').str.strip()
    return df

df = transform(df)
print(df.shape)
```
